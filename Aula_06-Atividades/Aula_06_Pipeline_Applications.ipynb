````python
import torch
import transformers
from transformers import pipeline

print("Versão do transformers:", transformers.__version__)
print("GPU disponível:", torch.cuda.is_available())

# Dispositivo: 0 = primeira GPU, -1 = CPU
DEVICE = 0 if torch.cuda.is_available() else -1
# Em GPU usamos meia precisão (float16) para economizar memória
DTYPE = torch.float16 if torch.cuda.is_available() else torch.float32
````

````python
#Modelo do Chat

MODELO_CHAT = "Qwen/Qwen3.5-0.8B"

chat = pipeline(
    "text-generation",
    model=MODELO_CHAT,
    dtype=DTYPE,
    device=DEVICE,
)
````

````python
mensagens = [
    {"role": "system", "content": "Você é um assistente prestativo que responde sempre em português do Brasil, de forma clara e concisa."},
    {"role": "user", "content": "Explique em duas frases o que é um modelo de linguagem."},
]

saida = chat(mensagens, max_new_tokens=150, do_sample=True, temperature=0.7)

# A saída traz a conversa completa; a resposta do modelo é a última mensagem
resposta = saida[0]["generated_text"][-1]["content"]
print(resposta)
````

## **Discussão**

Observamos que ao modificar as variáveis "do_sample" e "temperature" a resposta, ou seja, a saída do texto sempre é modificada para a abordagem do usuário "Explique em duas frases o que é um modelo de linguagem."
